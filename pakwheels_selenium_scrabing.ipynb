```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# Setup Chrome
options = Options()
options.add_argument("--start-maximized")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(options=options)

url = "https://www.pakwheels.com/used-cars/karachi/24857"
driver.get(url)

time.sleep(5)

# ---- AUTO SCROLL (load all cars) ----
last_height = driver.execute_script("return document.body.scrollHeight")

while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)

    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height

print("✅ Finished scrolling")

# ---- PARSE HTML ----
soup = BeautifulSoup(driver.page_source, "html.parser")
driver.quit()

# ---- CAR BLOCKS ----
cars = soup.find_all("div", class_="search-list clearfix ad-container page- pos-rel")

print(f"✅ Found {len(cars)} cars\n")

data = []

for car in cars:
    try:
        # Title
        title_tag = car.find("a", class_="car-name")
        title = title_tag.text.strip() if title_tag else "N/A"

        # Price
        price_tag = car.find("div", class_="price-details")
        price = price_tag.text.strip() if price_tag else "N/A"

        # Specs
        specs = car.find_all("li")
        year = mileage = fuel = engine = transmission = "N/A"

        for spec in specs:
            text = spec.text.strip()

            if "km" in text:
                mileage = text
            elif "Hybrid" in text or "Petrol" in text or "Diesel" in text:
                fuel = text
            elif "cc" in text:
                engine = text
            elif "Automatic" in text or "Manual" in text:
                transmission = text
            elif text.isdigit():
                year = text

        # Location
        location_tag = car.find("ul", class_="search-vehicle-info")
        location = location_tag.text.strip() if location_tag else "N/A"

        data.append([title, price, year, mileage, fuel, engine, transmission, location])

    except Exception as e:
        print("Error:", e)

# ---- PRINT RESULT ----
for d in data:
    print(d)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# ---------------- SETUP ----------------
options = Options()

options.add_argument("--start-maximized")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(options=options)

# ---------------- OPEN WEBSITE ----------------
url = "https://www.pakwheels.com/used-cars/karachi/24857"
driver.get(url)

# ==============================
# MANUAL SCROLL
# ==============================
print("\n✅ Page Opened")
print("👉 Ab browser mein khud scroll karo")
print("👉 Jab sari listings load ho jayein")
input("👉 ENTER press karo scraping start karne ke liye...")

# ---------------- GET PAGE SOURCE ----------------
html = driver.page_source

driver.quit()

# ---------------- PARSE HTML ----------------
soup = BeautifulSoup(html, "html.parser")

# Listing cards
cars = soup.find_all("div", class_="search-list")

print(f"\n✅ Found {len(cars)} cars\n")

# ---------------- EXTRACT DATA ----------------
all_cars = []

for car in cars:

    # ---------- TITLE ----------
    title_tag = car.find("h3")
    title = title_tag.get_text(strip=True) if title_tag else "N/A"

    # ---------- PRICE ----------
    price_tag = car.find("div", class_="price-details")
    price = price_tag.get_text(" ", strip=True) if price_tag else "N/A"

    # ---------- LOCATION ----------
    location = "N/A"

    location_tag = car.find("ul", class_="search-vehicle-info")

    if location_tag:
        location = location_tag.get_text(" ", strip=True)

    # ---------- DETAILS ----------
    year = "N/A"
    mileage = "N/A"
    fuel = "N/A"
    engine = "N/A"
    transmission = "N/A"

    details_tag = car.find("ul", class_="search-vehicle-info-2")

    if details_tag:

        details = details_tag.find_all("li")

        for item in details:

            text = item.get_text(strip=True)

            # Year
            if text.isdigit() and len(text) == 4:
                year = text

            # Mileage
            elif "km" in text:
                mileage = text

            # Fuel
            elif text in ["Petrol", "Hybrid", "Diesel", "Electric"]:
                fuel = text

            # Engine
            elif "cc" in text:
                engine = text

            # Transmission
            elif text in ["Automatic", "Manual"]:
                transmission = text

    # ---------- SAVE ----------
    all_cars.append({
        "Title": title,
        "Price": price,
        "Location": location,
        "Year": year,
        "Mileage": mileage,
        "Fuel": fuel,
        "Engine": engine,
        "Transmission": transmission
    })

# ---------------- PRINT RESULT ----------------
for idx, car in enumerate(all_cars, start=1):

    print("\n" + "=" * 60)
    print(f"Car #{idx}")

    for key, value in car.items():
        print(f"{key}: {value}")
```
